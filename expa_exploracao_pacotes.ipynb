## **Fluxo recomendado quando você não tem internet**

**PT:**
1. **Sei o módulo, mas não lembro o nome exato** > `dir(modulo)` - veja a lista completa
2. **Lembro uma parte do nome** > `[f for f in dir(m) if 'palavra' in f.lower()]`
3. **Encontrei o nome, mas não sei os parâmetros** > `inspect.signature(funcao)`
4. **Quero uma visão geral rápida e formatada** > `funcao+?` (Jupyter) ou `help(funcao)` (qualquer terminal)
5. **Quero entender o que cada parâmetro faz em detalhe** > `inspect.getdoc(funcao)` - texto puro, fácil de fatiar com `[:N]`
6. **Documentação confusa, quero ver o código real** > `inspect.getsource(funcao)`

> **PT:** A diferença entre os passos 4 e 5: `?` e `help()` entregam tudo de uma vez de forma legível; `inspect.getdoc()` retorna uma string - útil quando você quer exibir só um trecho ou processar o texto no código.

## **Exploração de Pacotes Python sem Internet**

**Por que isso é importante?**

Quando você está codando e não tem acesso à internet (em provas, ambientes restritos, ou simplesmente offline), o Python já traz embutido tudo que você precisa para *descobrir* o que um pacote pode fazer - sem precisar abrir nenhuma documentação externa.

---

**As duas ferramentas principais**

| Ferramenta | O que faz |
|---|---|
| dir(objeto) | Lista tudo que existe dentro de um módulo, classe ou objeto |
| inspect | Inspeciona funções: vê parâmetros, lê código-fonte, filtra tipos |

Pense no `dir()` como o *índice de um livro*, e no `inspect` como a *lupa* que permite ler os detalhes de cada capítulo.

```python
# Importações necessárias para o tutorial inteiro

import sklearn as sk    # pacote de machine learning
import pandas as pd     # pacote de manipulação de dados
import inspect          # módulo nativo do Python para introspecção
```

## 1. Listando o que existe dentro de um módulo

O ponto de partida é sempre `dir()`. Ele retorna uma lista com **tudo** que está disponível dentro do objeto passado: funções, classes, submódulos e até atributos internos. Os itens com `__` duplo são internos do Python - pode ignorá-los por enquanto.

**Dica:** Use `dir(modulo)` quando quiser ver *o que existe*, sem saber ainda o que procura.

```python
# dir() no módulo raiz: mostra os submódulos disponíveis no pacote
print("=== Submódulos do sklearn ===")
dir(sk)
```

```python
# Entrando em um submódulo específico para ver suas funções e classes
print("=== Conteúdo de sklearn.model_selection ===")
dir(sk.model_selection)
```

**Como analisar essa saída?**
- Ignore os itens que começam com `_` (são internos do Python)
- Nomes em `CamelCase` > geralmente **classes** (precisam ser instanciadas)
- Nomes em `snake_case` > geralmente **funções** (chamadas diretamente)

---
## 2. Pesquisando por palavra-chave dentro do módulo

Se você se lembra de um fragmento do nome (ex: "split", "grid", "cross"), pode filtrar a lista com uma *list comprehension* simples. É como fazer uma busca textual na documentação, mas direto no terminal.

**Dica:** Use `.lower()` na busca para não perder resultados por causa de maiúsculas/minúsculas.

```python
# Filtrar por palavra-chave: útil quando você lembra parte do nome

keyword = "split"   # troque aqui pela palavra que você quer buscar

resultado = [f for f in dir(sk.model_selection) if keyword in f.lower()]

print(f"=== Funções/classes com '{keyword}' em sk.model_selection ===")
resultado
```

```python
# Testando múltiplas palavras-chave de uma vez - útil para exploração rápida

for word in ["cross", "grid", "score", "fold"]:
    found = [f for f in dir(sk.model_selection) if word in f.lower()]
    if found:
        print(f"'{word}' >> {found}")
```

**Como usar essa saída?** Ao buscar "fold" você encontrará `KFold` e `StratifiedKFold`, revelando que há pelo menos duas variantes de validação cruzada disponíveis. Isso expande seu mapa mental do pacote.

---
## 3. Filtrando apenas funções ou apenas classes

O `inspect` permite filtrar pelo *tipo* de cada item. Isso é útil quando você quer ver apenas funções utilizáveis diretamente, ou apenas classes para instanciar, sem se distrair com todo o resto.

```python
# Filtrar apenas funções do módulo (excluindo classes, constantes, etc.)

so_funcoes = [
    f for f in dir(sk.model_selection)
    if inspect.isfunction(getattr(sk.model_selection, f))  # verifica se é função
    and not f.startswith("_")                              # ignora itens internos
]

print("Funções disponíveis em sk.model_selection:")
for f in so_funcoes:
    print(f"  - {f}")
```

```python
# Filtrar apenas classes (úteis quando você precisa instanciar um objeto)

so_classes = [
    f for f in dir(sk.model_selection)
    if inspect.isclass(getattr(sk.model_selection, f))  # verifica se é classe
    and not f.startswith("_")                           # ignora itens internos
]

print("Classes disponíveis em sk.model_selection:")
for c in so_classes:
    print(f"  - {c}")
```

**Como usar essa saída?**
- Precisa encaixar algo em um **pipeline**? > procure nas *classes* (`inspect.isclass`)
- Precisa de uma **operação direta** (ex: dividir dados)? > procure nas *funções* (`inspect.isfunction`)

---
## 4. Lendo a documentação de uma função ou classe

Depois de encontrar o que procura, você pode ler a *docstring* (documentação embutida) diretamente no Python. Há três formas principais:
| Método                   | Quando usar                                   |
|--------------------------|-----------------------------------------------|
| `objeto?`                | Visualização formatada - **apenas no Jupyter  |
| `help(objeto)`           | Funciona em qualquer terminal Python          |
| `inspect.getdoc(objeto)` | Quando quiser guardar o texto em uma variável |

```python
# Método 1: usando '?' no Jupyter (gera uma janela de ajuda formatada)

sk.model_selection.GridSearchCV?
```

```python
# Método 2: inspect.getdoc() - funciona em qualquer ambiente Python

doc = inspect.getdoc(sk.model_selection.GridSearchCV)

# Mostrando só as primeiras linhas para não poluir a saída
print(doc[:800])   # aumente esse número para ver mais
```

**Como analisar a docstring?** Ela costuma seguir esta estrutura:
1. **Descrição** - O que a função/classe faz
2. **Parameters** - O que ela aceita como entrada (com tipos)
3. **Returns / Attributes** - O que ela retorna ou expõe após o `fit`
4. **Examples** - Código de uso direto pronto para copiar

Vá direto para a seção *Parameters* quando não souber o que passar.

---
## 5. Inspecionando a assinatura de uma função (parâmetros)

Quando você só precisa saber **quais parâmetros** uma função aceita - sem ler toda a documentação - use `inspect.signature()`. É a forma mais rápida de descobrir o que passar como argumento.

```python
# Ver a assinatura completa da função (com os valores padrão de cada parâmetro)

sig = inspect.signature(sk.model_selection.GridSearchCV)
print("Assinatura completa || Full signature:")
list(sig.parameters.values())
```

```python
# Versão avançada: nome + valor padrão lado a lado - revela o que é obrigatório

print(f"{'Parâmetro':<25} {'Valor padrão':<30}")
print("-" * 55)

for name, param in inspect.signature(sk.model_selection.GridSearchCV).parameters.items():
    # Se não tem valor padrão, é um argumento obrigatório
    # If there's no default value, it's a required argument
    default = param.default if param.default is not inspect.Parameter.empty else "[!]  obrigatório"
    print(f"{name:<25} {str(default):<30}")
```

**Como usar essa saída?** Parâmetros marcados como *obrigatório* precisam ser sempre fornecidos - o código vai falhar sem eles. Os demais já têm valores padrão sensatos e você só precisa alterá-los para customizar o comportamento.

## 6. Explorando métodos de uma instância (objeto já criado)

Às vezes, você já tem um objeto instanciado e quer saber o que pode fazer *com ele*. `dir()` também funciona em instâncias!

```python
# Criar uma instância e inspecionar seus métodos e atributos públicos

from sklearn.linear_model import LinearRegression

model = LinearRegression()     # instanciando o objeto

# Listar apenas itens públicos (sem os internos com '__')
publics = [m for m in dir(model) if not m.startswith("_")]

print("Métodos e atributos públicos de LinearRegression():")
publics
```

```python
# Ler a documentação de um método específico do objeto

print(inspect.getdoc(model.fit))
```

**O que observar?** Objetos de ML geralmente expõem:
- `fit` - treinar o modelo
- `predict` - gerar previsões
- `score` - avaliar o desempenho
- Atributos com `_` no final (ex: `coef_`) - disponíveis **somente após** o `fit`

## 7. Exemplo prático: explorando pandas do zero

Vamos aplicar o fluxo completo para descobrir como usar `pd.read_csv` sem internet - um cenário muito comum.

```python
# Passo 1: descobrir todas as funções de leitura/escrita em pandas

funcoes_io = [f for f in dir(pd) if "read" in f.lower() or "to_" in f.lower()]
print("[PT] Funções de I/O em pandas")
funcoes_io
```

```python
# Passo 2: ver todos os parâmetros de pd.read_csv de uma vez

print("Assinatura de pd.read_csv:")
print(inspect.signature(pd.read_csv))
```

#### Exemplo prático: Quais são os parâmetros padrões e quais são obrigatórios em `sk.model_selection.GridSearchCV`?

```python
print(f"{'Parâmetro':<25} {'Valor padrão':<30}")
print("-" * 55)

module_name = sk.model_selection.GridSearchCV

for name, param in inspect.signature(module_name).parameters.items():
    # Se não tem valor padrão, é um argumento obrigatório
    # If there's no default value, it's a required argument
    default = param.default if param.default is not inspect.Parameter.empty else "[!]  obrigatório"
    print(f"{name:<25} {str(default):<30}")
```

```python
# Passo 3: ler a documentação para entender os parâmetros mais importantes

doc_csv = inspect.getdoc(pd.read_csv)
print(doc_csv[:1200])   # primeiras linhas
```

## 8. Bonus: ver o código-fonte de uma função

Em casos extremos, quando a documentação não é suficiente, o `inspect` permite ver o **código-fonte real** da função. Isso só funciona para código Python puro - funções compiladas em C não têm código-fonte acessível.

```python
# Ver o código-fonte de uma função Python pura (ex: train_test_split)

try:
    codigo = inspect.getsource(sk.model_selection.train_test_split)
    
    # Mostrar apenas as primeiras linhas para não poluir a saída
    linhas = codigo.split("\n")
    print("\n".join(linhas[:35]))
    print("\n... (código truncado)")

except TypeError:
    # Funções compiladas em C não têm código-fonte acessível via inspect
    print("[!]  Esta função é compilada em C - código-fonte não disponível.")
```
