## **Recommended workflow when you don't have internet**

1. **I know the module, but don't remember the exact name** > `dir(module)` - see the full list
2. **I remember part of the name** > `[f for f in dir(m) if 'word' in f.lower()]`
3. **Found the name, but don't know the parameters** > `inspect.signature(function)`
4. **I want a quick, formatted overview** > `function+?` (Jupyter) or `help(function)` (any terminal)
5. **I want to understand each parameter in detail** > `inspect.getdoc(function)` - plain text, easy to slice with `[:N]`
6. **Documentation is confusing, I want to see the real code** > `inspect.getsource(function)`

The difference between steps 4 and 5: `?` and `help()` deliver everything at once in a readable format; `inspect.getdoc()` returns a string - useful when you want to display only a portion or process the text in code.

## **Exploring Python Packages Without Internet**

**Why does this matter?**

When you're coding without internet access (in exams, restricted environments, or simply offline), Python already has everything built-in to *discover* what a package can do - without opening any external documentation.

---

**The two main tools**

| Tool        | What it does                                                           |
|-------------|------------------------------------------------------------------------|
| dir(object) | Lists everything inside a module, class, or object                     |
| inspect     | Inspects functions: views parameters, reads source code, filters types |

Think of `dir()` as the *table of contents* of a book, and `inspect` as the *magnifying glass* that lets you read the details of each chapter.

```python
# Required imports for the entire tutorial

import sklearn as sk    # machine learning package
import pandas as pd     # will be used in advanced examples
import inspect          # Python built-in introspection module        
```

## 1. Listing everything inside a module

The starting point is always `dir()`. It returns a list with **everything** available inside the passed object: functions, classes, submodules, and even internal attributes. Items with double `__` are Python internals - you can ignore them for now.

**Tip:** Use `dir(module)` when you want to see *what exists*, without knowing yet what you're looking for.

```python
# dir() on the root module: shows available submodules in the package
print("=== Submódulos do sklearn ===")
dir(sk)
```

```python
# Going into a specific submodule to see its functions and classes
print("=== Conteúdo de sklearn.model_selection ===")
dir(sk.model_selection)
```

**How to analyze this output?**
- Ignore items starting with `_` (they are Python internals)
- Names in `CamelCase` > usually **classes** (need to be instantiated)
- Names in `snake_case` > usually **functions** (called directly)

## 2. Searching by keyword inside a module

**EN:** If you remember a fragment of the name (e.g., "split", "grid", "cross"), you can filter the list with a simple *list comprehension*. It's like doing a text search in the documentation, but right in the terminal.

**Tip:** Use `.lower()` in the search so you don't miss results due to uppercase/lowercase.

```python
# Filter by keyword: useful when you remember part of the name

keyword = "split"   # replace with the word you want to search
                    
resultado = [f for f in dir(sk.model_selection) if keyword in f.lower()]

print(f"=== Funtions/classes with '{keyword}' in sk.model_selection ===")
resultado
```

```python
# Testing multiple keywords at once - useful for quick exploration

for word in ["cross", "grid", "score", "fold"]:
    found = [f for f in dir(sk.model_selection) if word in f.lower()]
    if found:
        print(f"'{word}' >> {found}")
```

**EN:** **How to use this output?** Searching "fold" finds `KFold` and `StratifiedKFold`, revealing there are at least two variants of cross-validation available. This expands your mental map of the package.

---
## 3. Filtering only functions or only classes

**EN:** `inspect` lets you filter by the *type* of each item. This is useful when you want to see only directly callable functions, or only classes to instantiate, without getting distracted by everything else.

```python
# Filter only functions from the module (excluding classes, constants, etc.)

just_functions = [
    f for f in dir(sk.model_selection)
    if inspect.isfunction(getattr(sk.model_selection, f))  # checks if it's a function
    and not f.startswith("_")                              # ignores internal items
]

print("Available functions in sk.model_selection:")
for f in just_functions:
    print(f"  - {f}")
```

```python
# Filter only classes (useful when you need to instantiate an object)

just_classes = [
    f for f in dir(sk.model_selection)
    if inspect.isclass(getattr(sk.model_selection, f))  # checks if it's a class
    and not f.startswith("_")                           # ignores internal items
]

print("Classes disponíveis em sk.model_selection:")
for c in just_classes:
    print(f"  - {c}")
```

**How to use this output?**
- Need something to fit in a **pipeline**? > look in the *classes* (`inspect.isclass`)
- Need a **direct operation** (e.g., split data)? > look in the *functions* (`inspect.isfunction`)

```python
# Method 1: using '?' in Jupyter (generates a formatted help window)

sk.model_selection.GridSearchCV?
```

```python
# Method 2: inspect.getdoc() - works in any Python environment

doc = inspect.getdoc(sk.model_selection.GridSearchCV)

# Showing only the first lines to keep the output clean
print(doc[:800])   # aumente esse número para ver mais / increase this number to see more
```

**PT:** **Como analisar a docstring?** Ela costuma seguir esta estrutura:
1. **Descrição** - O que a função/classe faz
2. **Parameters** - O que ela aceita como entrada (com tipos)
3. **Returns / Attributes** - O que ela retorna ou expõe após o `fit`
4. **Examples** - Código de uso direto pronto para copiar

Vá direto para a seção *Parameters* quando não souber o que passar.

**EN:** **How to analyze the docstring?** It usually follows this structure:
1. **Description** - What the function/class does
2. **Parameters** - What it accepts as input (with types)
3. **Returns / Attributes** - What it returns or exposes after `fit`
4. **Examples** - Ready-to-copy usage code

Go straight to the *Parameters* section when you don't know what to pass.

## 5. Inspecting a function's signature (parameters)

When you just need to know **which parameters** a function accepts - without reading all the documentation - use `inspect.signature()`. It's the fastest way to discover what to pass as arguments.

```python
# View the complete function signature (with default values for each parameter)

sig = inspect.signature(sk.model_selection.GridSearchCV)
print("Assinatura completa || Full signature:")
list(sig.parameters.values())
```

```python
# Advanced version: name + default value side by side - reveals what's required

print(f"{'Parameter':<25} {'Default value':<30}")
print("-" * 55)

for name, param in inspect.signature(sk.model_selection.GridSearchCV).parameters.items():
    # Se não tem valor padrão, é um argumento obrigatório
    # If there's no default value, it's a required argument
    default = param.default if param.default is not inspect.Parameter.empty else "[!]  required"
    print(f"{name:<25} {str(default):<30}")
```

**How to use this output?** Parameters marked as *required* must always be provided - the code will fail without them. The rest already have sensible defaults and you only need to change them to customize behavior.

## 6. Exploring methods of an instance (already created object)

Sometimes you already have an instantiated object and want to know what you can do *with it*. `dir()` works on instances too!

```python
# Create an instance and inspect its public methods and attributes

from sklearn.linear_model import LinearRegression

model = LinearRegression()     # instantiating the object

# List only public items (excluding internal ones with '__')
publics = [m for m in dir(model) if not m.startswith("_")]

print("Methods and atributes of LinearRegression():")
publics
```

```python
# Read the documentation of a specific method of the object

print(inspect.getdoc(model.fit))
```

**What to look for?** ML objects usually expose:
- `fit` - train the model
- `predict` - generate predictions
- `score` - evaluate performance
- Attributes ending with `_` (e.g., `coef_`) - available **only after** `fit`

## 7. Practical example: exploring pandas from scratch

Let's apply the full workflow to discover how to use `pd.read_csv` without internet - a very common scenario.

```python
# Step 1: discover all read/write functions in pandas

funcoes_io = [f for f in dir(pd) if "read" in f.lower() or "to_" in f.lower()]
print("I/O functions in pandas")
funcoes_io
```

```python
# Step 2: view all parameters of pd.read_csv at once

print("Signature of pd.read_csv:")
print(inspect.signature(pd.read_csv))
```

```python
print(f"{'Parameter':<25} {'Default value':<30}")
print("-" * 55)

module_name = pd.read_csv

for name, param in inspect.signature(module_name).parameters.items():
    # Se não tem valor padrão, é um argumento obrigatório
    # If there's no default value, it's a required argument
    default = param.default if param.default is not inspect.Parameter.empty else "[!]  required"
    print(f"{name:<25} {str(default):<30}")
```

```python
# Step 3: read the documentation to understand the most important parameters

doc_csv = inspect.getdoc(pd.read_csv)
print(doc_csv[:1200])   # first lines                     
```

## 8. Bonus: viewing the source code of a function

In extreme cases, when documentation isn't enough, `inspect` lets you see the **actual source code** of a function. This only works for pure Python code - functions compiled in C don't have accessible source code.

```python
# View the source code of a pure Python function (e.g., train_test_split)

try:
    codigo = inspect.getsource(sk.model_selection.train_test_split)
    
    # Show only the first lines to keep it clean
    linhas = codigo.split("\n")
    print("\n".join(linhas[:35]))
    print("\n... (code truncated)")

except TypeError:
    # Functions compiled in C don't have accessible source code via inspect
    print("[!]  This function is compiled in C - source code not available.")
```
